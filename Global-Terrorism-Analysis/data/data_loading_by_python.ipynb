```python
import csv

csv_path = r"C:\Users\MY PC\Desktop\Global-Terrorism-Analysis\data\globalterrorismdb_0718dist.csv"

with open(csv_path, "r", encoding="latin1", newline="") as f:
    reader = csv.reader(f)

    header = next(reader)

    for line_number, row in enumerate(reader, start=2):
        if len(row) != len(header):
            print("Problem found!")
            print("CSV line:", line_number)
            print("Columns found:", len(row))
            print("Expected:", len(header))
            print("Event ID:", row[0])
```

```python
import csv

input_file = r"C:\Users\MY PC\Desktop\Global-Terrorism-Analysis\data\globalterrorismdb_0718dist.csv"
output_file = r"C:\Users\MY PC\Desktop\Global-Terrorism-Analysis\data\gtd_import_ready.csv"

with open(input_file, "r", encoding="latin1", newline="") as infile, \
     open(output_file, "w", encoding="latin1", newline="") as outfile:

    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    header = next(reader)
    writer.writerow(header)

    skipped = 0
    imported_rows = 0

    for row in reader:
        if len(row) != len(header):
            skipped += 1
            continue

        writer.writerow(row)
        imported_rows += 1

print("Import-ready CSV created!")
print("Valid rows:", imported_rows)
print("Skipped rows:", skipped)
```
